```python
for col in df.select_dtypes(include='number'):
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

```python
columns = [
    'MonthlyIncome',
    'NumCompaniesWorked',
    'PerformanceRating',
    'StockOptionLevel',
    'TotalWorkingYears',
    'TrainingTimesLastYear',
    'YearsAtCompany',
    'YearsInCurrentRole',
    'YearsSinceLastPromotion',
    'YearsWithCurrManager'
]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    LL = Q1 - 1.5 * IQR
    UL = Q3 + 1.5 * IQR

    df[col] = df[col].clip(LL, UL)
```

```python
for col in df.select_dtypes(include='number'):
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(col)
    plt.show()
```

EDA 1: AGE DISTRIBUTIONS OF EMPLOYEES

```python
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.show()
```

EDA 2: Monthly Income vs Attrition

```python
sns.boxplot(x='Attrition', y='MonthlyIncome', data=df)
plt.title('Monthly Income vs Attrition')
plt.show()
```

EDA 3: Countplot – Overtime vs Attrition

```python
sns.countplot(x='OverTime', hue='Attrition', data=df)
plt.title('Overtime vs Attrition')
plt.show()
```

EDA 4: Correlation Heatmap

```python
columns = [
    'Age',
    'MonthlyIncome',
    'TotalWorkingYears',
    'YearsAtCompany',
    'JobSatisfaction'
]

sns.heatmap(df[columns].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

Feature 1: Income Per Job Level

```python
df['IncomePerJobLevel'] = df['MonthlyIncome'] / df['JobLevel']
```

Feature 2: Years Without Promotion Ratio

```python
df['PromotionRatio'] = df['YearsSinceLastPromotion'] / (df['YearsAtCompany'] + 1)
```

# **Key Insights**
1.   Overtime: Employees working overtime have a higher attrition rate (30.5%) than employees not working overtime (10.4%).
2.   Monthly Income: Employees who left generally had lower monthly incomes than those who stayed.

3.   Age: Employees who left were younger on average than employees who stayed.
4.   Work Experience: Investigate whether employees with fewer years at the company have higher attrition rates.
5.   Job Satisfaction: Investigate whether lower job satisfaction is associated with higher attrition.
